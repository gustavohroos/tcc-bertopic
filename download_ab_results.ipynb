```python
from google.cloud import bigquery

client = bigquery.Client()

timestamp = "2025-07-19"
```

```python
query = f"""
SELECT
  documentKey,
  doc,
  section,
  timestamp,
  section_name_list
FROM
  `gglobo-recplat-ml-hdg-prd`.`features`.`bertopic_v2`
WHERE
  TIMESTAMP_TRUNC(timestamp, DAY) >= TIMESTAMP("{timestamp}")
"""

query_job = client.query(query)

query_job.to_dataframe().to_csv("bertopic_v2_topics_globo.parquet", index=False)
```
