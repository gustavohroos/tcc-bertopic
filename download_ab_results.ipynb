```python
from google.cloud import bigquery

client = bigquery.Client()

timestamp_start = "2025-06-25"
timestamp_end = "2025-07-26"
```

```python
query = f"""
SELECT
  documentKey,
  doc,
  section,
  timestamp,
  section_name_list
FROM
  `gglobo-recplat-ml-hdg-prd`.`features`.`bertopic_v2`
WHERE
  TIMESTAMP_TRUNC(timestamp, DAY) >= TIMESTAMP("{timestamp_start}") AND TIMESTAMP_TRUNC(timestamp, DAY) <= TIMESTAMP("{timestamp_end}")
"""

query_job = client.query(query)

query_job.to_dataframe().to_csv(f"results/ab/bertopic_v2_topics_globo_{timestamp_start}_{timestamp_end}.csv", index=False)
```

```python
query = f"""
SELECT
  b.documentKey,
  b.doc,
  b.section,
  b.timestamp,
  b.section_name_list,
  d.issued
FROM
  `gglobo-recplat-ml-hdg-prd.features.bertopic` AS b
JOIN
  `gglobo-recplat-ml-hdg-prd.documents.rec_documents` AS d
ON
  b.documentKey = d.documentKey
WHERE
  TIMESTAMP_TRUNC(b.timestamp, DAY) >= TIMESTAMP("{timestamp_start}")
  AND TIMESTAMP_TRUNC(b.timestamp, DAY) <= TIMESTAMP("{timestamp_end}")
  AND b.alg_config = "915bf5c0f415fe59d5d63c0a832cbf307d560d542a0d03ac8513ad33791369ea"
"""

query_job = client.query(query)

query_job.to_dataframe().to_csv(f"results/ab/bertopic_v1_topics_globo_{timestamp_start}_{timestamp_end}.csv", index=False)
```
